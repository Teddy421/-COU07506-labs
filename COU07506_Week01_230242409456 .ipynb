```python
# Task 1.1: First Notebook

## Generation and Plotting of Elementary Sequences
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
n = np.arange(-5, 11)

delta = (n == 0).astype(float)
step = (n >= 0).astype(float)
ramp = n * step
expo = 0.8**n * step
```

```python
fig, axes = plt.subplots(1, 4, figsize=(13, 2.6))

for ax, x, title in zip(axes, [delta, step, ramp, expo],
                        ["δ(n)", "u(n)", "r(n)", "(0.8)^n u(n)"]):
    ax.stem(n, x, basefmt="k-")
    ax.set_title(title)
    ax.set_xlabel("n")

plt.tight_layout()
plt.show()
```

Task 1.2 — Change the Parameters

```python

a_values = [0.5, -0.8, 1.1]

fig, axes = plt.subplots(1, 3, figsize=(12, 3))

for ax, a in zip(axes, a_values):
    x = a**n * step
    ax.stem(n, x, basefmt="k-")
    ax.set_title(f"a = {a}")
    ax.set_xlabel("n")
    ax.set_ylabel("Amplitude")

plt.tight_layout()
plt.show()
```

```python

n = np.arange(0, 41)

cos1 = np.cos(2 * np.pi * n / 8)
cos2 = np.cos(2 * np.pi * n / 7.5)

fig, axes = plt.subplots(1, 2, figsize=(12, 3))

axes[0].stem(n, cos1, basefmt="k-")
axes[0].set_title("cos(2πn/8)")
axes[0].set_xlabel("n")
axes[0].set_ylabel("Amplitude")

axes[1].stem(n, cos2, basefmt="k-")
axes[1].set_title("cos(2πn/7.5)")
axes[1].set_xlabel("n")
axes[1].set_ylabel("Amplitude")

plt.tight_layout()
plt.show()
```

The signal \(\cos(2\pi n/8)\) repeats exactly, while \(\cos(2\pi n/7.5)\) does not repeat exactly. This is because a discrete-time cosine signal is periodic only when its period is a rational number.

```python
# Task 1.3: Working With Audio

## Recording, Loading and Plotting an Audio Signal
```

```python
import os
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import wavfile
from IPython.display import Audio

fname = r"C:\Users\HomePC\Documents\Sound Recordings\my_voice.wav"

if os.path.exists(fname):
    fs, x = wavfile.read(fname)

    if x.ndim > 1:
        x = x.mean(axis=1)

    x = x.astype(float)
    x = x / np.max(np.abs(x))

    t = np.arange(len(x)) / fs

    print(f"fs = {fs} Hz")
    print(f"Duration = {len(x)/fs:.2f} seconds")
    print(f"Samples = {len(x)}")

    plt.figure(figsize=(12, 4))
    plt.plot(t, x)
    plt.xlabel("Time (s)")
    plt.ylabel("Amplitude")
    plt.title("My Voice Recording")
    plt.show()

else:
    print("Audio file not found. Check the file path.")
```

```python
Audio(x, rate=fs)
```

### 1. What is the sampling rate of your phone’s recording? How many numbers are needed to store one minute of it in mono?

The sampling rate of my phone recording is **48,000 Hz**.

The number of samples needed to store one minute of mono audio is calculated as:

\[
N = f_s \times 60
\]

\[
N = 48,000 \times 60
\]

\[
N = 2,880,000 \text{ samples}
\]

Therefore, **2,880,000 numbers** are needed to store one minute of mono audio.

### 2. Why is a stem plot more appropriate than a continuous line for x(n)?

A stem plot is more appropriate because x(n) is a discrete-time signal. It displays individual sample values clearly, while a continuous line plot may incorrectly suggest that the signal exists at every point in time.

### 3. For a = 1.1, the exponential grows. Which property of a system would such a signal violate if it were an impulse response?

It would violate the **BIBO stability property (Bounded-Input, Bounded-Output stability)**.

This is because the exponential impulse response grows continuously and is not absolutely summable. Therefore, the system would be unstable.
